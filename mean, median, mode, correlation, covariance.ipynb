# Mean, Median, Mode

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
dataset= sns.load_dataset("titanic")
```

```python
dataset
```

```python
dataset.head(3)
```

```python
dataset.isnull().sum() 
```

```python
dataset['age'].fillna(dataset['age'].mean(), inplace=True)
```

```python
dataset.isnull().sum()
```

```python
dataset['age']
```

```python
np.percentile(dataset['age'], 25), np.percentile(dataset['age'], 75)      # percentile 25 and 75
```

```python
np.percentile(dataset['age'], 0)   # percentile 0
```

```python
dataset['age'].min()    # minimum value of the dataset in age column
```

```python
dataset['age'].max()         # maximum value of the dataset in age column
```

```python
dataset['age'].mean()        # mean value of the dataset in age column
```

```python
dataset['age'].median()      # median value of the dataset in age column
```

```python
dataset['age'].mode()         # mode value of the dataset in age column
```

```python
dataset.describe()
```

```python
sns.boxplot(x='age', data=dataset)
```

```python
sns.boxplot(x='fare', data=dataset)
```

# Correlation

```python
dataset1= sns.load_dataset("tips")
```

```python
dataset1
```

```python
dataset1.head(3)
```

```python
dataset1.isnull().sum()
```

```python
dataset1.info()
```

```python
data_corr =dataset1.select_dtypes(['float64','int64']).corr()     # correlation between the numerical columns in the dataset1
data_corr
```

```python
data_cov= dataset1.select_dtypes(['float64','int64']).cov()      # covariance between the numerical columns in the dataset1  
data_cov 
```

```python
sns.heatmap(data_corr, annot=True)
plt.show()
```

```python
sns.heatmap(data_cov, annot=True)
plt.show()
```
